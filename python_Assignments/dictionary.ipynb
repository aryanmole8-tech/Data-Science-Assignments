```python
# 3. Flatten nested dictionary[cite: 13]
def flatten_dict(d, parent_key=''):
    items = []
    for k, v in d.items():
        new_key = f"{parent_key}.{k}" if parent_key else k
        if isinstance(v, dict):
            items.extend(flatten_dict(v, new_key).items())
        else:
            items.append((new_key, v))
    return dict(items)

d = {'student': {'name': 'Amit', 'marks': {'Python': 90}}}
print(flatten_dict(d))

# 4. Recursively merge two nested dictionaries[cite: 13]
def deep_merge(d1, d2):
    res = d1.copy()
    for k, v in d2.items():
        if k in res and isinstance(res[k], dict) and isinstance(v, dict):
            res[k] = deep_merge(res[k], v)
        else:
            res[k] = v
    return res

D1 = {'a': 1, 'b': {'x': 10}}
D2 = {'b': {'y': 20}, 'c': 3}
print(deep_merge(D1, D2))

# 5. Update inventory quantities[cite: 13]
inventory = {'Pen': 10, 'Book': 5}
transactions = [('Pen', -3), ('Book', 4), ('Pencil', 6)]
for item, qty in transactions:
    inventory[item] = inventory.get(item, 0) + qty
print(inventory)

# 6. Leaderboard with ranks[cite: 13]
scores = {'A': 95, 'B': 82, 'C': 95, 'D': 70}
grouped = {}
for player, score in scores.items():
    grouped.setdefault(score, []).append(player)

rank = 1
for score in sorted(grouped.keys(), reverse=True):
    players = ", ".join(grouped[score])
    print(f"Rank {rank}: {players} ({score})")
    rank += 1

# 7. Reverse dependency map[cite: 13]
tasks = {'Deploy': ['Test'], 'Test': ['Build'], 'Build': ['Code']}
rev = {}
for task, prereqs in tasks.items():
    for p in prereqs:
        rev.setdefault(p, []).append(task)
print(rev)

# 8. Highest-scoring subject per student[cite: 14]
scores = {
    'Amit': {'Python': 90, 'ML': 85},
    'Neha': {'Python': 88, 'ML': 94}
}
for student, sub_scores in scores.items():
    top_sub = max(sub_scores, key=sub_scores.get)
    print(f"{student} -> {top_sub} ({sub_scores[top_sub]})")

# 9. Add two sparse matrices[cite: 14]
A = {(0, 0): 2, (1, 1): 3}
B = {(0, 0): 5, (0, 1): 4}
C = A.copy()
for k, v in B.items():
    C[k] = C.get(k, 0) + v
print(dict(sorted(C.items())))

# 10. LRU cache simulation using insertion order[cite: 14]
capacity = 3
requests = ['A', 'B', 'C', 'A', 'D']
cache = {}
for r in requests:
    if r in cache:
        del cache[r]
    elif len(cache) >= capacity:
        cache.pop(next(iter(cache)))
    cache[r] = True
print(f"Final cache order = {list(cache.keys())}")
```
