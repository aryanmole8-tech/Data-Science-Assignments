```python
# 1. Numbers 1 to 10[cite: 15]
print(*(range(1, 11)))

# 2. Even numbers 1 to 20[cite: 15]
print(*(range(2, 21, 2)))

# 3. Odd numbers 1 to 20[cite: 15]
print(*(range(1, 20, 2)))

# 4. Reverse order 10 to 1[cite: 15]
print(*(range(10, 0, -1)))

# 5. Multiplication table[cite: 15]
num = 5
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")

# 6. Sum of numbers 1 to n[cite: 15]
n = 5
print(f"Sum = {sum(range(1, n + 1))}")

# 7. Factorial using range[cite: 15]
n = 5
fact = 1
for i in range(1, n + 1):
    fact *= i
print(f"Factorial = {fact}")

# 8. Squares from 1 to 10[cite: 16]
print(*[i**2 for i in range(1, 11)])

# 9. Cubes from 1 to 10[cite: 16]
print(*[i**3 for i in range(1, 11)])

# 10. Count numbers divisible by 7 (1 to 100)[cite: 16]
print(f"Count = {len(range(7, 101, 7))}")

# 11. Numbers 1 to 50 except multiples of 5[cite: 16]
print(*[i for i in range(1, 51) if i % 5 != 0])

# 12. Fibonacci series up to n terms[cite: 16]
n = 7
a, b = 0, 1
fib = []
for _ in range(n):
    fib.append(a)
    a, b = b, a + b
print(*fib)

# 13. ASCII values of uppercase letters[cite: 16]
for i in range(65, 91):
    print(f"{chr(i)} = {i}")

# 14. List 100 to 0 step -10[cite: 16]
print(list(range(100, -1, -10)))

# 15. Leap years between two years[cite: 16]
start, end = 2000, 2020
leaps = [y for y in range(start, end + 1) if (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0)]
print(*leaps)

# 16. Star pattern using nested range[cite: 17]
rows = 4
for i in range(1, rows + 1):
    print("*" * i)
```
