```python
sentence = "python is easy and python is powerful"
words = sentence.split()
seen, duplicates = set(), set()
for word in words:
    if word in seen:
        duplicates.add(word)
    seen.add(word)
print(f"Duplicate words = {duplicates}")
```

Section C - Advanced

```python
nums = {100, 4, 200, 1, 3, 2}
longest = 0
best_seq = []
for n in nums:
    if n - 1 not in nums:
        curr = n
        seq = []
        while curr in nums:
            seq.append(curr)
            curr += 1
        if len(seq) > longest:
            longest = len(seq)
            best_seq = seq
print(f"Sequence = {best_seq}\nLength = {longest}")
```

```python
nums = [2, 7, 11, 15, 1, 8]
target = 9
seen, pairs = set(), set()
for n in nums:
    diff = target - n
    if diff in seen:
        pairs.add(tuple(sorted((n, diff))))
    seen.add(n)
print(f"Pairs = {pairs}")
```

```python
s = {1, 2, 3}
lst = list(s)
power_set = [[]]
for elem in lst:
    power_set += [curr + [elem] for curr in power_set]
print(f"Number of subsets = {len(power_set)}")
print(f"Subsets: {[set(x) for x in power_set]}")
```

```python
row = [5, 3, 4, 6, 7, 8, 9, 1, 2]
print(f"Valid Sudoku row = {len(row) == 9 and set(row) == set(range(1, 10))}")
```

```python
lst = [1, 2, 1, 3, 4, 2, 3]
k = 4
distinct_counts = [len(set(lst[i:i+k])) for i in range(len(lst) - k + 1)]
print(f"Distinct counts = {distinct_counts}")
```

```python
graph = {
    'A': {'B', 'C', 'D'},
    'B': {'A', 'C', 'E'}
}
common = graph['A'] & graph['B']
print(f"Common neighbors of A and B = {common}")
```

```python
friends = {
    'A': {'B', 'C'},
    'B': {'A', 'D'},
    'C': {'A', 'D', 'E'}
}
user = 'A'
fof = set()
for f in friends[user]:
    fof |= friends[f]
recommendations = fof - friends[user] - {user}
print(f"Recommendations for A = {recommendations}")
```

```python
lst = [1, 2, 2, 4, 5]
n = 5
seen = set()
duplicate = next(x for x in lst if x in seen or seen.add(x))
missing = (set(range(1, n + 1)) - set(lst)).pop()
print(f"Duplicate = {duplicate}\nMissing = {missing}")
```

```python
universe = {1, 2, 3, 4, 5}
groups = [{1, 2}, {3, 4}, {5}]
is_disjoint = all(groups[i].isdisjoint(groups[j]) for i in range(len(groups)) for j in range(i + 1, len(groups)))
union_all = set().union(*groups)
print(f"Valid disjoint partition = {is_disjoint and union_all == universe}")
```

```python
universe = {1, 2, 3, 4, 5}
sets = {'S1': {1, 2, 3}, 'S2': {2, 4}, 'S3': {3, 4, 5}}
covering_pair = None
for k1 in sets:
    for k2 in sets:
        if k1 != k2 and sets[k1] | sets[k2] == universe:
            covering_pair = (k1, k2)
            break
print(f"Smallest covering pair = {covering_pair}")
```
