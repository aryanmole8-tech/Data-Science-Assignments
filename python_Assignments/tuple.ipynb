```python
matrix = ((1,2,3),(4,5,6),(7,8,9))
num_rows = len(matrix)
num_cols = len(matrix[0])

sums_list = []
for i in range(num_cols):
    current_sum = 0
    for j in range(num_rows):
        current_sum += matrix[i][j]
    sums_list.append(current_sum)
column_sums = tuple(sums_list)
print(f"Column-wise Sum : {column_sums}")
```

```python
#8
tuple1 = (1,2,3,4,5)
n = 2
if len(tuple1) > n:
    shift = n % len(tuple1)
    result = tuple1[shift:] + tuple1[:shift]
else:
    result = tuple1
print(f"Tuple after shifting by {n} positions : {result}")
```

```python
#9
tup = (1,2,5,3,4,6,7,2)
max_sub = ()
current_sub = ()

if len(tup) > 0:
    max_sub = (tup[0],)
    current_sub = (tup[0],)

    for i in range(1, len(tup)):
        if tup[i] > tup[i-1]:
            current_sub += (tup[i],)
        else:
            if len(current_sub) > len(max_sub):
                max_sub = current_sub
            current_sub = (tup[i],)

    if len(current_sub) > len(max_sub):
        max_sub = current_sub

print(f"Subsequence = {max_sub}")
print(f"Length = {len(max_sub)}")
```

```python
#10
tuple_a = (1, 3, 5, 7)
tuple_b = (2, 4, 6, 8, 10)

i, j = 0, 0
merged_list = []

while i < len(tuple_a) and j < len(tuple_b):
    if tuple_a[i] < tuple_b[j]:
        merged_list.append(tuple_a[i])
        i += 1
    else:
        merged_list.append(tuple_b[j])
        j += 1

while i < len(tuple_a):
    merged_list.append(tuple_a[i])
    i += 1

while j < len(tuple_b):
    merged_list.append(tuple_b[j])
    j += 1

result = tuple(merged_list)

print("Merged Sorted Tuple:", result)
```
