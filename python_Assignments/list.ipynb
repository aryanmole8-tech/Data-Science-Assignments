```python
l2 = [8,3,15,1,9]
smallest = min(l2)
largest = max(l2)
print(f"Smallest = {smallest}")
print(f"Largest = {largest}")
```

```python
l3 = [1,2,3,4,5,6]
even_num = []
odd_num = []
for i in l3:
    if i%2 == 0:
        even_num.append(i)
    else:
        odd_num.append(i)
print(f"Even Numbers : {even_num}")
print(f"Odd Numbers : {odd_num}")
```
