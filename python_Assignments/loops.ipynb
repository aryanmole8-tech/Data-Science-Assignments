17.  Write a Python program to check whether a number is palindrome or not.

```python
num = int(input("Enter Number : "))
print(f"Number : {num}")
palindrome = num
rev_num = 0
while num > 0:
    d = num % 10
    rev_num = rev_num * 10 + d
    num = num // 10
print(f"Reverse of Number : {rev_num}")
if palindrome == rev_num:
    print("The number is a palindrome.")
else:
    print("The number is not a palindrome.")
```

18. Write a Python program to find frequency of each digit in a given integer

```python
num = int(input("Enter Number : "))
print(f"Number : {num}")
cnt = {}
while num > 0:
    digit = num % 10
    if digit in cnt:
        cnt[digit] += 1
    else:
        cnt[digit] = 1
    num = num // 10
print(cnt)
```

23. Write a Python program to find HCF (GCD) of two numbers.

```python
n1 = int(input("Enter First Number : "))
n2 = int(input("Enter Second Number : "))
hcf = 1
for i in range(1, min(n1, n2) + 1):
    if n1 % i == 0 and n2 % i == 0:
        hcf = i
print(f"HCF of {n1} and {n2} is : {hcf}")
```

26. Write a Python program to print all Prime numbers between 1 to n.

```python
n = int(input("Enter value of n : "))
print(f"Prime numbers from 1 to {n} are : ")
while n >= 2:
    is_prime = True
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            is_prime = False
            break
    if is_prime:
        print(n, end=" ")
    n -= 1
```

28. Write a Python program to check whether a number is Armstrong number 
or not.

```python
n = int(input("Enter Number : "))
print(f"Number : {n}")
power = len(str(n))
temp = n
armstrong_sum = 0
while temp > 0:
    digit = temp % 10
    armstrong_sum += digit ** power
    temp //= 10
if armstrong_sum == n:
    print(f"{n} is an Armstrong number.")
else:
    print(f"{n} is not an Armstrong number.")
```

29. Write a Python program to print all Armstrong numbers between 1 to n

```python
n = int(input("Enter value of n : "))
total_sum = 0
print(f"Armstrong numbers between 1 to {n} are :")
for num in range(1, n + 1):
    power = len(str(num))
    temp = num 
    armstrong_sum = 0

    while temp > 0:
        digit = temp % 10
        armstrong_sum += digit ** power
        temp //= 10

    if num == armstrong_sum:
        print(num, end = " ")
        total_sum += num

print()
print(f"Sum of Armstrong numbers between 1 to {n} : {total_sum}")
```

30. Write a Python program to print Fibonacci series up to n terms.

```python
n = int(input("Enter Number : "))
a, b = 0, 1
count = 0

if n <= 0:
    print("Please enter positive integer")
else:
    print("Fibonacci series : ")
    while count < n:
        print(a, end = " ")
        next_term = a + b
        a = b
        b = next_term
        count += 1

print()
```
