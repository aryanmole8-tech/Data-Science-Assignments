```python
# 1 : Write a Python program to find maximum between two numbers.
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
print(num1, ",", num2)
if num1 > num2:
    print(f"Maximum number :{num1}")
else:
    print(f"Maximum number :{num2}")
```

```python
# Q2 : Write a Python program to find the Maximum among three numbers
n1 = int(input("Enter Num1 : "))
n2 = int(input("Enter Num2 : "))
n3 = int(input("Enter Num3 : "))

if (n1>n2 and n1>n3):
    print(f"{n1} is Maximum")
elif(n2>n1 and n2>n3):
    print(f"{n2} is Maximum")
else:
    print(f"{n3} is Maximum")
```

```python
# Q6. Write a Python program to check whether a year is leap year or not.
yr = int(input("Enter the Year : "))
if yr % 4 == 0 and yr % 100 != 0 or yr % 400 == 0:
    print(f"{yr} is a Leap Year")
else:
    print(f"{yr} is not a Leap Year")
```

```python
# Q7. Write a Python program to check whether a character is alphabet or not.
ch = input("Enter the Character : ")
if (ch.isalpha()):
    print(f"{ch} is an Alphabet")
else:
    print(f"{ch} is not an Alphabet")
```

```python
# Q8. Write a Python program to input any alphabet and check whether it is vowel or consonant.
ch = input("Enter the Character : ")
if (ch.isalpha() and ch.lower() in ['a', 'e', 'i', 'o', 'u']):
    print(f"{ch} is a Vowel")
else:
    print(f"{ch} is a Consonant")
```

```python
# Q9. Write a Python program to check whether a character is uppercase or lowercase alphabet.
ch = input("Enter the Character : ")
if (ch.isalpha()):
    if ch == ch.lower():
        print("Character is Lowercase")
    elif ch == ch.upper():
        print("Character is Uppercase")
else:
    print("Not Character")
```

```python
# Q10. Write a Python program to count total number of notes in given amount.
amount = int(input("Enter the Amount : "))
notes = [2000, 500 , 200, 100, 50, 20, 10, 5 ,2, 1]
total_notes = 0
for note in notes:
    count = amount // note
    if count > 0:
        print(note, ":", count)
        total_notes += count
        amount = amount % note
print("Total Notes :", total_notes)
```

```python
# 12. Write a Python program to input all sides of a triangle and check whether the triangle is valid or not.
s1 = int(input("Enter Side 1 : "))
s2 = int(input("Enter Side 2 : "))
s3 = int(input("Enter Side 3 : "))
if (s1 + s2 > s3) and (s2 + s3 > s1) and (s1 + s3 > s2):
    print("Triangle is Valid")
else:
    print("Triangle is Not Valid")
```

```python
# 13. Write a Python program to check whether the triangle is equilateral, isosceles or scalene triangle.
a = int(input("Enter side a:"))
b = int(input("Enter side b:"))
c = int(input("Enter side c:"))

if (a == b and b == c):
    print("Triangle is Equivalent")
elif (a == b or b == c or a == c):
    print("Triangle is Isosceles")
else:
    print("Triangle is Scelene")
```

```python
# 14. Write a Python program to calculate profit or loss. Input is selling cost and actual cost.
a_cost = int(input("Enter the Actual Cost"))
s_cost = int(input("Enter the Selling Cost"))

if (s_cost > a_cost):
    profit = s_cost - a_cost
    print(f"Profit : {profit}")

elif (a_cost > s_cost):
    loss = a_cost - s_cost
    print(f"Loss : {loss}")

else:
    print("No Profit, No Loss")
```

```python
# 15. Write a Python program to input marks of five subjects Physics, Chemistry,
# Biology, Mathematics and Computer. Calculate percentage and grade 

m1 = int(input("Enter Marks of Physics : "))
m2 = int(input("Enter Marks of Chemistry : "))
m3 = int(input("Enter Marks of Biology : "))
m4 = int(input("Enter Marks of Mathematics : "))
m5 = int(input("Enter Marks of CS : "))

percentage = ((m1+m2+m3+m4+m5)/5)
print(percentage)

if percentage >= 90:
    print("Grade A")
elif percentage >= 80:
    print("Grade B")
elif percentage >= 70:
    print("Grade C")
elif percentage >= 60:
    print("Grade D")
elif percentage >= 40:
    print("Grade E")
elif percentage < 40 :
    print("Grade F")
else:
    print("Fail")

```

```python
# Write a Python program to input basic Python salary of an employee and 
# calculate its Gross salary according to following:
# Basic Salary <= 10000 : HRA = 20%, DA = 80%
# Basic Salary <= 20000 : HRA = 30%, DA = 90%
# Basic Salary > 20000 : HRA = 35%, DA = 95%

salary = float(input("Enter the Basic Salary : "))
if salary <= 10000 :
    hra = salary * 0.20
    da = salary * 0.80

elif salary <= 20000:
    hra = salary * 0.30
    da = salary * 0.90

else:
    hra = salary * 0.35
    da = salary * 0.95

gross_salary = salary + hra + da
print(f"Gross Salary : {gross_salary}")
```

```python
# Write a Python program to input electricity unit charges and calculate total 
# electricity bill according to the given condition:
# For first 50 units Rs. 0.50/unit
# For next 100 units Rs. 0.75/unit
# For next 100 units Rs. 1.25/unit
# For unit above 250 Rs. 1.50/unit
# An additional surcharge of 17% is added to the bill

units = int(input("Enter the Electricity Unit Charges : "))
bill_amount = 0

if units <= 50:
    bill_amount = units * 0.50

elif units <= 150:
    bill_amount = (50 * 0.50) + ((units - 50) * 0.75)

elif units <= 250:
    bill_amount = (50 * 0.50) + (100 * 0.75) + ((units - 150) * 1.25)

else :
    bill_amount = (50 * 0.50) + (100 * 0.75)  + (100 * 1.25) + ((units - 250) * 50)

surcharges = bill_amount * 0.17
total_bill = bill_amount + surcharges

print(f"Base Bill Amount : Rs. {bill_amount : .2f}")
print(f"Surcharges : Rs. {surcharges : .2f}")
print(f"Total Electricity Charges : Rs. {total_bill : .2f}")
```
