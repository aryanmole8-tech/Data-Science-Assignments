```python
# Q2 : Write a Python program to find the length of a string without using len().
l = input("Enter a string: ")
print(l)
cnt = 0
for i in l:
    cnt += 1
print(f"Length of String : {cnt}")
```

```python
# Q3 :  Write a Python program to count the number of vowels and consonants in a given string.
s = input("Enter the String: ")
print(s)
vowels = 0
consonants = 0
for i in s:
    if i == 'a' or i == 'e' or i == 'i' or i == 'o' or i == 'u':
        vowels += 1
    else:
        consonants += 1
print(f"Number of vowels: {vowels}")
print(f"Number of consonants: {consonants}")
```

```python
# Q4: Write a Python program to count uppercase letters, lowercase letters, digits, and special characters in a
# string.
s = input("Enter a string: ")
upper_letters = 0
lower_letters = 0
digits = 0
special_char = 0
for i in s:
    if i.isupper():
        upper_letters += 1
    elif i.islower():
        lower_letters += 1
    elif i.isdigit():
        digits += 1
    else:
        special_char += 1
print(f"Uppercase : {upper_letters}")
print(f"Lowercase : {lower_letters}")
print(f"Digits : {digits}")
print(f"Special Character : {special_char}")
```

```python
# Q5 : Write a Python program to check whether a given string is a palindrome.
s = input("Enter the String : ")
if s == s[::-1]:
    print("The String is Palindrome")
else:
    print("The String is not Palindrome")
```

```python
# Q6 : Write a Python program to reverse a string without using slicing
s = input("Enter the String : ")
rev = ""
for i in s:
    rev = i + rev
print(f"Reverse of the String : {rev}")
```

```python
# Q8 :  Write a Python program to count the frequency of a given character in a string.
s = input("Enter the string : ")
c = input("Enter the character to count : ")
print(s)
print(f"Character to count : {c}")
cnt = 0
for i in s:
    if i == c:
        cnt += 1
print(f"Count of the character : {cnt}")
```

```python
# Q9 :  Write a Python program to find the first occurrence and last occurrence of a character in a string.
s = input("Enter the string : ")
c = input("Enter the character to find : ")
print(s)
print(f"Character to find : {c}")
first_occurrence = -1
last_occurrence = -1
for i, char in enumerate(s):
    if char == c:
        if first_occurrence == -1:
            first_occurrence = i
        last_occurrence = i
if first_occurrence != -1:
    print(f"First occurrence of {c} is at index : {first_occurrence}")
    print(f"Last occurrence of {c} is at index : {last_occurrence}")
```

```python
# Q11 : Write a Python program to check whether two strings are anagrams of each other.
s = input("Enter Str1 : ")
t = input("Enter Str2 : ")
print(s,t, sep = ',')
if sorted(s) == sorted(t):
    print("The strings are anagrams of each other.")
else:
    print("The strings are not anagrams of each other.")
```

```python
# Q12 : Write a Python program to find the most frequently occurring character in a string.
s = input("Enter the String : ")
print(s)
frequency = {}
for i in s:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1
max_char = ""
max_count = 0
for i , count in frequency.items():
    if count > max_count:
        max_count = count
        max_char = i
print("The most frequently occurring character is :", max_char)
```

```python
# Q13 :  Write a Python program to remove duplicate characters from a string while preserving the original order.
s = input("Enter the String : ")
print(s)
res = []
for i in s:
    if i not in res:
        res.append(i)
print("".join(res))

```

```python
# Q14 : Write a Python program to count the number of words in a sentence.
s = input("Enter the Sentence : ")
print(s)
cnt = 0
for i in s.split():
    cnt += 1
print(f"Number of words in the sentence : {cnt}")
```

```python
# Q15 : Write a Python program to find the longest word in a sentence.
s = input("Enter the Sentence : ")
print(s)
longest_word = ""
for i in s.split():
    if len(i) > len(longest_word):
        longest_word = i
print(f"The longest word in the sentence is : {longest_word}")
```

```python
# Q16 :  Write a Python program to capitalize the first letter of every word in a sentence without using title().
s = input("Enter the Sentence : ")
print(s)
capitalized = []
for i in s.split():
    capitalized.append(i.capitalize())
print(" ".join(capitalized))
```
